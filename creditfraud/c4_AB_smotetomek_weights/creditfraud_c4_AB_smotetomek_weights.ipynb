Protocol (identical for every dataset and config, defined in `common/`): 5x2 repeated stratified
outer CV; inside each outer training fold a 20-candidate random search with
3-fold inner CV, optimising macro-F1; 7 classifiers
(LR, LinearSVM, GNB, DT, RF, XGB, LGBM). Imputation, scaling, resampling and weighting are all fitted
inside the training data only.
